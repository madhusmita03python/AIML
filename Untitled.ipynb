```python
import csv

totalexpenses = []
budget = 0.0
file_name = "expenses.csv"
remaining_budget = 0.0

def load_expenses():
    global totalexpenses
    totalexpenses = []
    try:
        file = open(file_name, 'r')
        lines = file.readlines()
        file.close()

        for line in lines[1:]:
            parts = line.strip().split(',')

            if len(parts) != 4:
                continue

            date, category, amount_str, description = parts

            try:
                amount = float(amount_str)
            except ValueError:
                continue

            expense = {
                'date': date,
                'category': category,
                'amount': amount,
                'description': description
            }

            totalexpenses.append(expense)

        print(f"\nLoaded {len(totalexpenses)} expenses from file.")

    except FileNotFoundError:
        print("\nNo saved expense file found. Starting fresh.")

def add_expense():
    global remaining_budget
    if remaining_budget <= 0:
        exit()
    date = input("Enter the date (YYYY-MM-DD): ")
    category = input("Enter category (e.g., Food, Travel): ")
    try:
        amount = float(input("Enter the amount spent: "))
    except ValueError:
        print("Invalid amount. Please enter a number.")
        return
    description = input("Enter a brief description: ")

    expense = {
        'date': date,
        'category': category,
        'amount': amount,
        'description': description
    }
    totalexpenses.append(expense)
    print("Expense added successfully.")

def view_expenses():
    if not totalexpenses:
        print("No expenses in the list")
        return
    print("\n--- Expense List ---")
    for i, expense in enumerate(totalexpenses, 1):
        if all(k in expense for k in ['date', 'category', 'amount', 'description']):
            print(f"{i}. {expense['date']}  {expense['category']}  {expense['amount']:.2f}  {expense['description']}")
        else:
            print(f"{i}. Incomplete expense entry (skipped)")

def track_budget():
    global budget, remaining_budget
    totalSpent = 0
    if budget == 0.0:
        budget = float(input("\n Enter your monthly budget: ₹"))
    if budget < 0:
        print("Invalid amount")
        return
    for expense in totalexpenses:
        totalSpent += float(expense['amount'])
    print("\nTotal spent amount: ₹", totalSpent)

    remaining_budget = budget - totalSpent
    if remaining_budget < 0:
        print("Exceeded your budget!")
    else:
        print(f"You have ₹{remaining_budget:.2f} left for the month.")

def save_expenses():
    file = open(file_name, 'w')
    file.write('date,category,amount,description\n')
    for expense in totalexpenses:
        line = f"{expense['date']},{expense['category']},{expense['amount']},{expense['description']}\n"
        file.write(line)
    file.close()
    print("\nExpenses saved successfully in CSV format.")

def menu():
    while True:
        print("\n--- Personal Expense Tracker ---")
        print("1. Add expense")
        print("2. View expenses")
        print("3. Track budget")
        print("4. Save expenses")
        print("5. Exit")
        choice = input("Enter your choice (1-5): ")

        if choice == '1':
            add_expense()
        elif choice == '2':
            view_expenses()
        elif choice == '3':
            track_budget()
        elif choice == '4':
            save_expenses()
        elif choice == '5':
            save_expenses()
            print("!! Exiting the program !!")
            break
        else:
            print("!! Invalid option !! Choose the correct choice")

load_expenses()
menu()
```
